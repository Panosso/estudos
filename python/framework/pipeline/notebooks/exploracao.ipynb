```python
import json
import csv
```

```python
path_json = "../data_raw/dados_empresaA.json"

with open(path_json, "r") as file:
    dados_json = json.load(file)

for produto in dados_json:
    print(produto['Nome do Produto'])
    print("\n")
```

```python
path_csv = "../data_raw/dados_empresaB.csv"

with open(path_csv, "r") as file:
    #spamreader is the dafault name for that variable which reads the CSV file
    #Using dictionary reader to read CSV file and return a dict type
    spamreader = csv.DictReader(file, delimiter=',')
    for row in spamreader:
        print(row['Nome do Item'])
        print("\n")

```
